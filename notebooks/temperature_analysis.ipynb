```python
import pandas as pd

df = pd.read_csv("../data/temperature.csv", skiprows=8)
print(df.head())
```

```python
df.columns = df.columns.str.strip()
```

```python
df["time"] = pd.to_datetime(df["time"])
df["temperature_c"] = df["mean_M2SDNXSLV_5_12_4_T2MMEAN"] - 273.15

print(df[["time", "temperature_c"]].head())
print(df.dtypes)
```

```python
df["year"] = df["time"].dt.year

yearly_avg = df.groupby("year")["temperature_c"].mean().reset_index()

print(yearly_avg)
```

```python

import matplotlib.pyplot as plt

plt.figure(figsize=(12, 5))
plt.plot(yearly_avg["year"], yearly_avg["temperature_c"], marker="o")
plt.title("Annual Average Temperature (2000–2025)")
plt.xlabel("Year")
plt.ylabel("Temperature (°C)")
plt.grid(True)
plt.show()
```
